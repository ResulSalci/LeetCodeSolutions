```python
#1291. Sequential Digits
def sequentialDigits(low: int, high: int):
    result = []
    d = "123456789"
    len_low = len(str(low))
    len_high = len(str(high))

    if len_low > len_high:
        return result
        
    for i in range (len_low, len_high + 1):
        for j in range(0, len(d) - i +1):
            num = int(d[j:j + i])
            if low <= num <= high:
                result.append(num)
    return result
```
